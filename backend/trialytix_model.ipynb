```python
df = pd.read_csv("data/AERO-BirdsEye-Data.csv")
df.head()
```

```python
df["Status"].value_counts()
```

```python
# Keep only completed and failed trials
df_filtered = df[df["Status"].isin(["Completed", "Terminated", "Withdrawn", "Suspended"])]

# Create binary target variable
df_filtered["Success"] = df_filtered["Status"].apply(
    lambda x: 1 if x == "Completed" else 0
)

# Check new distribution
df_filtered["Success"].value_counts()

```

```python
df_filtered["Success"].value_counts()
```

```python
# Select relevant columns
features = ["Phase", "Enrollment", "Start_Year", "Sponsor", "Condition"]

X = df_filtered[features]
y = df_filtered["Success"]

X.head()
```

```python
X["Enrollment"].describe()

```

```python
# One-hot encode categorical features
X_encoded = pd.get_dummies(X, columns=["Phase", "Sponsor", "Condition"], drop_first=True)

X_encoded.head()

```

```python
X_encoded.shape
```

```python
from sklearn.model_selection import train_test_split

# Split data
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y, 
    test_size=0.2, 
    random_state=42,
    stratify=y  # very important for imbalance
)

X_train.shape, X_test.shape
```

```python
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier(
    n_estimators=200,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

model.fit(X_train, y_train)
```

```python
from sklearn.metrics import classification_report, roc_auc_score

y_pred = model.predict(X_test)
y_proba = model.predict_proba(X_test)[:, 1]

print(classification_report(y_test, y_pred))
print("ROC-AUC:", roc_auc_score(y_test, y_proba))
```

```python
import pandas as pd

# Get feature importance
importances = pd.Series(model.feature_importances_, index=X_encoded.columns)

# Top 10 important features
top_features = importances.sort_values(ascending=False).head(10)

top_features
```

```python
# Example: take first test sample
sample = X_test.iloc[0:1]

pred_prob = model.predict_proba(sample)[0][1]

print("Predicted Success Probability:", round(pred_prob * 100, 2), "%")
```
